```python
import pandas as pd

# 1. Define your list of target ISINs
target_isins = ["INF179K01UT0", "INF109K012R0"]  # Add as many ISINs as needed

# 2. Clean whitespace on ISIN column and filter using .isin()
filtered_df = df[df['isin'].astype(str).str.strip().isin(target_isins)].copy()

if filtered_df.empty:
    print("No matching records found for the given ISINs.")
else:
    # 3. Ensure 'date' is datetime format and sort chronologically
    filtered_df['date'] = pd.to_datetime(filtered_df['date'])
    filtered_df = filtered_df.sort_values('date')

    # 4. Group by ISIN and pick the last row (latest date) for each group
    latest_navs = filtered_df.groupby('isin', as_index=False).last()

    # Display clean result with selected columns
    result = latest_navs[['isin', 'scheme_name', 'date', 'nav']]
    print(result)
```
